## API Data Transformation and Cleansing

This notebook demonstrates:
- API extraction using JSON data
- At least 5 transformation/cleansing steps
- Ethical implications discussion
- Human-readable transformed dataset


```

import requests
import pandas as pd
import numpy as np
from difflib import get_close_matches

pd.set_option('display.max_columns', None)

url = "https://openlibrary.org/search.json?q=data+science"

response = requests.get(url)
data = response.json()

print("Status Code:", response.status_code)
print("Number of records returned:", len(data['docs']))
```


# Step #1 – Normalize JSON into a DataFrame

The nested API JSON response is transformed into a tabular pandas DataFrame
to improve readability and support downstream transformations.


```

records = data['docs'][:100]

df = pd.json_normalize(records)

selected_columns = [
    'title',
    'author_name',
    'first_publish_year',
    'edition_count',
    'language',
    'publisher'
]

df = df[selected_columns]

df.head()
```


# Step #2 – Replace Headers

Column names are converted into cleaner, human-readable labels.


```

df.columns = [
    'Title',
    'Author',
    'First Publish Year',
    'Edition Count',
    'Language',
    'Publisher'
]

df.head()
```


# Step #3 – Fix Missing Values

Null or missing values are identified and replaced with placeholder values.


```

df['Author'] = df['Author'].apply(lambda x: x[0] if isinstance(x, list) and len(x) > 0 else 'Unknown')
df['Language'] = df['Language'].apply(lambda x: ', '.join(x) if isinstance(x, list) else 'Unknown')
df['Publisher'] = df['Publisher'].apply(lambda x: x[0] if isinstance(x, list) and len(x) > 0 else 'Unknown')

df['First Publish Year'] = df['First Publish Year'].fillna(df['First Publish Year'].median())

df.head()
```


# Step #4 – Identify Outliers and Bad Data

Publication years outside realistic historical ranges are removed.


```

df = df[(df['First Publish Year'] >= 1900) & (df['First Publish Year'] <= 2026)]

df.head()
```


# Step #5 – Find and Remove Duplicates

Duplicate records based on Title and Author combinations are removed.


```

before = len(df)

df = df.drop_duplicates(subset=['Title', 'Author'])

after = len(df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
```


# Step #6 – Fix Casing and Inconsistent Values

Text formatting is standardized using title casing.


```

df['Title'] = df['Title'].str.title()
df['Author'] = df['Author'].str.title()
df['Publisher'] = df['Publisher'].str.title()

df.head()
```


# Step #7 – Conduct Simple Fuzzy Matching

Similar publisher names are grouped together using fuzzy matching.


```

publishers = df['Publisher'].dropna().unique().tolist()

def fuzzy_match(value, choices):
    matches = get_close_matches(value, choices, n=1, cutoff=0.85)
    return matches[0] if matches else value

df['Publisher Standardized'] = df['Publisher'].apply(lambda x: fuzzy_match(x, publishers))

df[['Publisher', 'Publisher Standardized']].head()
```


# Final Human-Readable Dataset


```

final_df = df.reset_index(drop=True)

print(final_df.head(20))

final_df.info()
```


# Ethical Implications

## What changes were made to the data?
The API JSON data was normalized into a DataFrame, missing values were handled,
duplicate rows were removed, text casing was standardized, and publisher names
were standardized using fuzzy matching.

## Are there any legal or regulatory guidelines for your data or project topic?
OpenLibrary provides publicly accessible metadata. However, attribution and
responsible use of API data should still be maintained.

## What risks could be created based on the transformations done?
Fuzzy matching may incorrectly group different publishers together.
Removing records with missing values could unintentionally remove valid data.

## Did you make any assumptions in cleaning/transforming the data?
The notebook assumed publication years outside the range 1900–2026 were invalid.
It also assumed the first publisher and first author in each list were the most relevant.

## How was your data sourced/verified for credibility?
The data was collected directly from the OpenLibrary public API, a widely used
digital library metadata source.

## Was your data acquired in an ethical way?
Yes. The API data is publicly accessible and was collected through standard API requests.

## How would you mitigate any ethical implications you have identified?
Additional validation checks and manual review could be implemented for fuzzy matching
and outlier detection steps.
