```python
# 1. Load the dataset as a Pandas data frame
import pandas as pd
df = pd.read_csv("./data/video_game_sales.csv", encoding = "utf-8")
```

```python
# 2. Display the first ten rows of data
ten = df.head(10)
print(ten)
```

```python
df = df.sort_values(by = "Critic_Score", ascending = False)
df.head(6)
html(4, "4. Find the top five games by critic score", "Tony Hawk's Pro Skater 2, SoulCalibur, Grand Theft Auto IV, Grand Theft Auto V, Super Mario Galaxy") 
```

```python
html(4, "5. Find the number of video games in the dataframe in each genre", """Racing = 1249<br>
Simulation = 874<br>
Action = 3370<br>
Platform = 888<br>
Shooter = 1323<br>
Puzzle = 580<br>
Role-Playing = 1500<br>
Strategy = 683<br>
Sports = 2348<br>
Adventure = 1303<br>
Fighting = 849<br>
Misc = 1750<br>""")
genre = df["Genre"]
unique = set(list(genre))

for x in unique:
    df2 = df[df["Genre"] == f"{x}"]
    print(f"{x} = {len(df2)}<br>")
```

```python
# Find the first five games in the data frame on the SNES platform
columns = ["Platform"]
for x in columns:
    print(set(list(df[x])))

snes = df.loc[df["Platform"] == "SNES", "Name"]
snes.head(5)

html(4, "6. Find the first five games in the data frame on the SNES platform", "Super Mario World, Super Mario All-Stars, Donkey Kong Country, Super Mario Kart, Street Fighter II")
```

```python
# Find the five publishers with the highest total global sales. Note: You will need to calculate the total global sales for each publisher.
publishers = set(list(df["Publisher"]))

publisher_global_sales = {}
for x in publishers:
    df2 = df.loc[df["Publisher"] == f"{x}"]
    global_sale = sum(df2["Global_Sales"])
    publisher_global_sales[x] = global_sale

df3 = pd.DataFrame(list(publisher_global_sales.items()))
df3.columns = ["Publisher", "Global_Sales"]
df3 = df3.sort_values(by = "Global_Sales", ascending = False)
df3.head(5)

html(4, "7. Find the five publishers with the highest total global sales. Note: You will need to calculate the total global sales for each publisher.",
     """1. Nintendo (1,788.81 sales)<br>
     2. Electronic Arts (1,116.96 sales)<br>
     3. Activision (731.16 sales)<br>
     4. Sony Computer Entertainment (606.48 sales)<br>
     5. Ubisoft (471.61 sales)""")
```

```python
html(4, "8. Create a new column in the data frame that calculates the percentage of global sales from North America", "")
df["Percent_from_NA (%)"] = round((df["NA_Sales"]/df["Global_Sales"] * 100),2)
df = df.drop(columns = "Percent_NA")
df.head(5)
```

```python
html(4, "9. Find the number of NaN (missing values in each column)")

columns = list(df.columns)
df = df.fillna("")
for x in columns:
    nan = df[df[x] == ""]
    print(f"{x} = {len(nan)}")

```
